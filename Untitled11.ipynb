```python
%%writefile main.cpp

#include <iostream>
#include <iomanip>
#include <vector>
#include <cmath>
#include <random>
#include <numeric>
#include <limits>

const double MU = 20.0;
const int PACKET_COMPLETIONS = 10000;
const int REPLICATIONS = 5;

double exponentialRandom(std::mt19937 &gen, double mean) {
    std::uniform_real_distribution<> dis(0.0, 1.0);
    return -mean * log(1 - dis(gen));
}

double simulateSingleReplication(double lambda,
                                 std::mt19937 &interArrivalGen,
                                 std::mt19937 &serviceTimeGen) {

    double currentTime = 0.0;
    double nextArrivalTime = 0.0;
    double nextDepartureTime = std::numeric_limits<double>::max();

    int queueSize = 0;
    double totalQueueLength = 0.0;
    double lastEventTime = 0.0;

    int packetCompletions = 0;

    while (packetCompletions < PACKET_COMPLETIONS) {

        if (nextArrivalTime < nextDepartureTime) {

            currentTime = nextArrivalTime;

            totalQueueLength +=
                queueSize * (currentTime - lastEventTime);

            lastEventTime = currentTime;

            queueSize++;

            nextArrivalTime =
                currentTime +
                exponentialRandom(interArrivalGen, 1 / lambda);

            if (queueSize == 1) {
                nextDepartureTime =
                    currentTime +
                    exponentialRandom(serviceTimeGen, 1 / MU);
            }

        } else {

            currentTime = nextDepartureTime;

            totalQueueLength +=
                queueSize * (currentTime - lastEventTime);

            lastEventTime = currentTime;

            queueSize--;
            packetCompletions++;

            if (queueSize > 0) {
                nextDepartureTime =
                    currentTime +
                    exponentialRandom(serviceTimeGen, 1 / MU);
            } else {
                nextDepartureTime =
                    std::numeric_limits<double>::max();
            }
        }
    }

    return totalQueueLength / currentTime;
}

void runSimulation(double lambda, double rho) {

    std::vector<std::mt19937> interArrivalGenerators(REPLICATIONS);
    std::vector<std::mt19937> serviceTimeGenerators(REPLICATIONS);

    for (int i = 0; i < REPLICATIONS; ++i) {
        interArrivalGenerators[i].seed(i * 2 + 1);
        serviceTimeGenerators[i].seed(i * 2 + 2);
    }

    std::vector<double> avgQueueLengths(REPLICATIONS);

    for (int r = 0; r < REPLICATIONS; ++r) {
        avgQueueLengths[r] =
            simulateSingleReplication(
                lambda,
                interArrivalGenerators[r],
                serviceTimeGenerators[r]
            );
    }

    double mean =
        std::accumulate(
            avgQueueLengths.begin(),
            avgQueueLengths.end(),
            0.0
        ) / REPLICATIONS;

    double sampleVariance = 0.0;

    for (double length : avgQueueLengths) {
        sampleVariance +=
            std::pow(length - mean, 2);
    }

    sampleVariance /= (REPLICATIONS - 1);

    double stddev = std::sqrt(sampleVariance);

    double halfWidth =
        1.96 * stddev / std::sqrt(REPLICATIONS);

    double accuracy =
        (halfWidth / mean) * 100;

    std::cout << std::fixed
              << std::setprecision(4)
              << rho << "\t"
              << mean << "\t["
              << mean - halfWidth << ", "
              << mean + halfWidth << "]\t\t"
              << halfWidth << "\t\t"
              << sampleVariance << "\t";

    std::cout << std::setprecision(2)
              << accuracy << "%\n";
}

int main() {

    std::vector<double> rhoValues = {
        0.1, 0.4, 0.6, 0.65, 0.7,
        0.72, 0.74, 0.78, 0.79, 0.8
    };

    std::cout << "\nSimulation of M/M/1 Queue\n";
    std::cout << "=======================\n";

    std::cout
        << "rho\tE[N]\tConfidence Interval (95%)"
        << "\tHalf-width\tSample Variance\tAccuracy\n";

    std::cout
        << "-----------------------------------------------------------------------------------\n";

    for (double rho : rhoValues) {

        double lambda = rho * MU;

        runSimulation(lambda, rho);
    }

    return 0;
}
```
